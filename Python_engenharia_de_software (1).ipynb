```python
def cadastrar_notas():
  notas = []

  while True:
    entrada = input("Digite uma nota (ou'sair' para finalizar): ")

    if entrada.lower() == 'sair':
      break

    try:
      nota = float(entrada)
      if 0 <= nota <= 10:
        notas.append(nota)
      else:
        print("Por favor, digite uma nova nota válida entre 0 e 10.")
    except ValueError:
      print("Entrada inválida! Digite um número ou a palavra 'sair'.")

  return notas

def calcular_media(notas):
  if len(notas) == 0:
    return 0.0

  media = sum(notas) / len(notas)
  return media

def determinar_situacao(media):
    if media >= 7.0:
        return "Aprovado"
    else:
        return "Reprovado"

def exibir_relatorio():
  print("---INÍCIO DO CADASTRO DE NOTAS---")
  lista_de_notas = cadastrar_notas()

  if len(lista_de_notas) == 0:
    print("Nenhuma nota foi cadastrada.")
    return

  media_final = calcular_media(lista_de_notas)
  situacao_final = determinar_situacao(media_final)

  print("\n=================================")
  print("        RELATÓRIO FINAL           ")
  print("==================================")
  print(f"Notas inseridas: {lista_de_notas}")
  print(f"Média do aluno: {media_final:.2f}")
  print(f"Situação do aluno: {situacao_final}")
  print("==================================")
exibir_relatorio()
```
